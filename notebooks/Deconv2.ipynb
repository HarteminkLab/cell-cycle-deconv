```python
def plot_raw_data(gene_name):
    orfs = pd.read_csv('datasets/orfs.csv').set_index('orf_name')
    mapping = pd.read_csv('datasets/gene.lst', names=['orf_name']).reset_index().set_index('orf_name')
    mapping = mapping.join(orfs)

    gene = mapping[mapping['name'] == gene_name].iloc[0]
    idx = gene['index']
    g_data = pd.read_csv('datasets/wt1.txt', sep='\t', header=None)
    g = g_data.loc[idx]
    plt.plot(g)
    plt.title(f'{gene.name} - {gene["name"]}')
    print(gene.name)
```

```python
plot_raw_data('SSK22')
```
