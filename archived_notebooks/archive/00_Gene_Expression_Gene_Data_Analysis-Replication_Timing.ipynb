Repeat, modified version of the analysis we performed for the chromatin raw data. We are checking to see if the gene expression data follows the same pattern of changes relative to nearby origins of replication. Potentially at the 30 minute and 20 minute timepoints in the first and second replicates respectively.


```python
# Preamble, notebook setup and imports

%load_ext autoreload
%autoreload 2
%matplotlib inline
%config InlineBackend.figure_format = 'retina'

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from cc_src.raw_transcription_analysis import TranscriptionDataAnalysis

analysis = TranscriptionDataAnalysis(window=2000)
```

```python
from src.timer import Timer

timer = Timer()

for rep in [1, 2]:
    for chrom in [3, 6, 10]:
        print(f"Creating heatmap for rep.{rep}, chrom.{chrom}, {timer.get_time()}")
        analysis.load_replicate_chromosome(replicate=rep, chrom=chrom)
        analysis.calculate_bins()
        fig = analysis.plot_heatmap()
        save_path = f'output/2024-03-12_Transcription_Comparison_Origins_Timing/chr{chrom}_rep{rep}.png'
        plt.savefig(save_path, dpi=150)
        plt.close(fig)
        print(f"Saved to {save_path} - {timer.get_time()}")
```
