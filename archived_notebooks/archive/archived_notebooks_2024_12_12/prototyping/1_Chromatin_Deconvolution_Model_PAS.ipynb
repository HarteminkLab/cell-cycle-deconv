```python
from src.global_config import GlobalConstants
from glob import glob
import numpy as np

pas_f_files = glob('output/deconvolve_sharedg1_g0066_11x11_2024_09_12/chromatin/*_f_*')
len(pas_f_files)
```

```python

f_imgs = None
orf_names = []
n = len(cluster_orfs)

i = 0
for file in pas_f_files:

    orf_name = file.split('/')[-1].split('_')[2]
    
    if not orf_name in cluster_orfs: continue
            
    try:
        pas_f = np.load(file)
    except:
         continue

    orf_names.append(orf_name)

    if f_imgs is None:
        f_imgs = np.zeros((n, *pas_f.shape))

    f_imgs[i] = pas_f
    i += 1

f_imgs = f_imgs.reshape((n, -1, *GlobalConstants.IMAGE_SHAPE))
```

```python
loaded_f_index_df = pd.DataFrame({'loaded_index': np.arange(len(orf_names))},
    index=orf_names)
loaded_f_index_df
```

```python
from src.geneset import get_deconvolved_geneset

geneset = get_deconvolved_geneset()
loaded_geneset = loaded_f_index_df.join(geneset, how='inner')

flip_crick_indices = loaded_geneset[loaded_geneset.strand == '-'].loaded_index
flip_crick_indices
```

```python
strand_corrected_f_imgs = f_imgs.copy()
strand_corrected_f_imgs[flip_crick_indices] = np.flip(f_imgs[flip_crick_indices], axis=3)
```

```python
strand_corrected_f_imgs.shape
```

```python
clusters_df = pd.read_csv('output/deconvolve_sharedg1_g0066_11x11_2024_08_13/'
                          'clustering/clusters.csv')[['cluster', 'orf_name']]
clusters_df = clusters_df.set_index('orf_name')
```

```python
from src.helpers import proportion_indices
from src.config import load_configs_by_config_type

config, _ = load_configs_by_config_type('shared')

g1_indices = config.get_Hpositions_for_phase('CG1')
s_indices = config.get_Hpositions_for_phase('S')
g2m_indices = config.get_Hpositions_for_phase('G2M')

sel_g1_indices = proportion_indices(g1_indices, [0.2, 0.8])
sel_s_indices = proportion_indices(s_indices, [0.2, 0.8])
sel_g2m_indices = proportion_indices(g2m_indices, [0.2, 0.8])

phase_indices = np.concatenate([
    sel_g1_indices,
    sel_s_indices,
    sel_g2m_indices
])
```

```python
from src.figure_configs import FiguresConfig

joined_clustered_orfs = loaded_f_index_df.join(clusters_df, how='inner')
joined_clustered_orfs = joined_clustered_orfs.reset_index()\
    .set_index(['cluster', 'index']).sort_index()

clusters = np.arange(1, 11)

def plot_im(ax, im):
    ax.imshow(im, cmap='magma_r', interpolation='none', 
        vmax=3, origin='lower', aspect='auto')
    ax.set_xticks([])
    ax.set_yticks([])

for cluster in clusters:
    cluster = 4
    fig, axs = plt.subplots(2, 3, figsize=(11, 3))    
    axs = np.array(axs).T.flatten()
    
    f_gene_indices = joined_clustered_orfs.loc[cluster].loaded_index
    clustered_f = strand_corrected_f_imgs[f_gene_indices]

    clustered_f_mean = clustered_f.mean(axis=0)
    
    for i in range(len(phase_indices)):
        ax = axs[i]
        plot_im(axs[i], clustered_f_mean[phase_indices[i]])
        ax.set_ylabel(f"{i+1}", fontsize=FiguresConfig.FIG_LABEL_FONTSIZE, 
            rotation=0, ha='right', labelpad=9)
        if i == 0: ax.set_title('Shared G1', fontsize=FiguresConfig.FIG_TITLE_FONTSIZE)
        elif i == 2: ax.set_title('S', fontsize=FiguresConfig.FIG_TITLE_FONTSIZE)
        elif i == 4: ax.set_title('G2/M', fontsize=FiguresConfig.FIG_TITLE_FONTSIZE)

    plt.suptitle(f"Cluster {cluster}, n={len(clustered_f)}")
    plt.subplots_adjust(top=0.80)
    break
```
