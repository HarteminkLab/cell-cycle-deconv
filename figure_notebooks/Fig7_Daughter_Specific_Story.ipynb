```python
from src.sgd import get_orfname
from src.config import load_default_chrom_configs, retrieve_phase_index_ticks

config, _ = load_default_chrom_configs()

gene_groups = [('blue', ['CLN1', 'CLN2', 'CLN3']),
        ('green', ['CLB5', 'CLB6']),
        ('red', ['CLB3', 'CLB4']),
        ('orange', ['CLB1', 'CLB2'])]

plt.figure(figsize=(8, 4))
for color, genes in gene_groups:
    
    avg = np.zeros(len(config.b_indices()))
    for gene in genes:
        orf_name = get_orfname(gene)
        clb1 = fig_daughter.expression_processor.expression_data.loc[orf_name]
        values = clb1[config.b_indices()].values
        values = (values - values.min()) / (values.max() - values.min())
        plt.plot(values, color=color, lw=5)
        
        x = values.argmax()
        
        if color == 'red':
            ha = 'left'
        elif color == 'orange':
            ha = 'right'
        else:
            ha = 'center'

        plt.text(x, 1.025, gene, color=color, fontsize=16, ha=ha)
        break

phase_ticks, edge_ticks, tick_labels = retrieve_phase_index_ticks('b', config, True)


ax = plt.gca()
ax.set_xticks(phase_ticks)
ax.set_xticklabels(tick_labels, rotation=0, ha='center', fontsize=13, fontweight='demi')
ax.set_xticks(edge_ticks, minor=True)
ax.set_yticks([])
ax.set_ylim(0, 1.15)
ax.tick_params(axis='x', which='major', length=0)
ax.tick_params(axis='x', which='minor', length=10)
ax.set_xlim(edge_ticks[0], edge_ticks[-1])
```
