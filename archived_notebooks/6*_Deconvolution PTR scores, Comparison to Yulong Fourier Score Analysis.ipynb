# PTR Comparison to Yulong's Fourier Score Analysis


```python
# Preamble, notebook setup and imports

%load_ext autoreload
%autoreload 2
%matplotlib inline
%config InlineBackend.figure_format = 'retina'

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Yulong's fourier analysis
# He has RNA-seq, promoter occupancy, and gene body occupancy fourier scores
# That we can compare against.

rna_f_scores = pd.read_csv('data/reference_data/yulongs_2023_Table_S1_rna_f_score.csv')
prom_tf_f_scores = pd.read_csv('data/reference_data/yulongs_2023_Table_S1_promoter_tf_score.csv')
gb_nuc_f_scores = pd.read_csv('data/reference_data/yulongs_2023_Table_S1_gene_body_nuc_score.csv')

rna_f_scores = rna_f_scores.set_index('Gene_ID')
```

```python
plt.figure(figsize=(12, 3))

plt.subplot(1, 3, 1)
plt.hist(rna_f_scores.Fourier_score, bins=50, color=plt.get_cmap('tab10')(4))
plt.title("RNA F-scores")

plt.subplot(1, 3, 2)
plt.hist(prom_tf_f_scores.Fourier_score, bins=50, color=plt.get_cmap('tab10')(1))
plt.title("Promoter TF F-scores")

plt.subplot(1, 3, 3)
plt.hist(gb_nuc_f_scores.Fourier_score, bins=50, color=plt.get_cmap('tab10')(0))
plt.title("Gene Body Nucleosome F-scores")

0
```
