```python

x = np.random.normal(10, 10, 1000)
y = np.random.normal(10, 10, 1000)
x[0:10]
```

```python
import statsmodels.api as sm

bw = [1, 1]

kde = sm.nonparametric.KDEMultivariate(data=[x, y], var_type='cc', bw=bw)
z = kde.pdf([x, y])

ax = plt.gca()
ax.scatter(x, y, c=z, s=10, cmap='magma')
```
