```python
# Preamble, notebook setup and imports

%load_ext autoreload
%autoreload 2
%matplotlib inline
%config InlineBackend.figure_format = 'retina'

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from cc_src.ComputePlusOne import ComputePlusOne
from src.timer import Timer

def compute_plus_ones(replicate):
    
    timer = Timer()
    chromatin_model = ComputePlusOne()
    gene_plus_ones = chromatin_model.geneset[[]].copy()
    gene_plus_ones['+1'] = None

    # Loop through all genes and find the +1 location
    count = 0
    for orf_name, _ in gene_plus_ones.iterrows():
        chromatin_model.load_mnase_gene(orf_name, replicate=replicate, log=False)

        # Compute +1 nucleosome for each gene
        gene_plus_ones.loc[orf_name, '+1'] = chromatin_model.computed_plus_one
        
        if count % 100 == 0:
            print(f"{count+1}/{len(gene_plus_ones)} - {timer.get_time()}")

        count += 1

    return gene_plus_ones
```
