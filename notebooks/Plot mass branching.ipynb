# Cell Cycle Branch Plotting

The purposes of this notebook is to plot an example of the cell cycle branching process across multiple cohorts as a function of experimental time.

What we hope to see, is clarity on the mu0 parameter. We are attempting to explore the idea of offsetting mu0 in the positive direction for alpha factor arrested experiments.

```python
from pysrc.plot_branch_timing import BranchPlotter

branch_plotter = BranchPlotter()
```

```python
branch_plotter.plot_figure(t=0)
0
```

```python
i = 0
for t in np.arange(0, 245, 5):
    fig = branch_plotter.plot_figure(t=t)
    plt.savefig(f'output/{i}.png', dpi=150)
    plt.cla()
    plt.clf()
    plt.close(fig)
    i += 1
```
