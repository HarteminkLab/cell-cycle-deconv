```python
def plot_deciles_of_average_tx(fig_nucleosomes_deciles, metric, summary_func):
    orfs_to_p1 = fig_nucleosomes_deciles.integrated_data.copy()
    orfs_to_p1 = orfs_to_p1[['matched_nuc_id_p1']].dropna()
    orfs_to_p1.matched_nuc_id_p1 = orfs_to_p1.matched_nuc_id_p1.astype(int)
    orfs_to_p1 = orfs_to_p1.reset_index().set_index('matched_nuc_id_p1')

    from src.config import load_default_chrom_configs
    config1, _ = load_default_chrom_configs()
    bins = np.linspace(0, 13, 50)

    for i in range(10):
        first_decile = fig_nucleosomes_deciles.all_cyclicity_groups['plus_one'][metric]\
            [i].nuc_id.values
        first_decile_orfs = orfs_to_p1.loc[first_decile].ORF

        first_decile_avg = summary_func(fig_nucleosomes_deciles\
            .expression_processor.expression_data.loc[
            first_decile_orfs][config1.t_indices()], axis=1)

        plt.boxplot(first_decile_avg, positions=[i+1], vert=False,
                   showfliers=False)#,
                    #flierprops=dict(markersize=1, marker='o', markerfacecolor='black'))
    plt.ylim(10.5, 0.5)
    plt.xlabel("Average transcript level")
    plt.title(metric.title())
    
plt.figure(figsize=(9, 4))
plt.subplot(1, 3, 1)
plot_deciles_of_average_tx(fig_nucleosomes_deciles, 'positioning', np.mean)
plt.ylabel("Cyclicity decile")
    
plt.subplot(1, 3, 2)
plot_deciles_of_average_tx(fig_nucleosomes_deciles, 'occupancy', np.mean)
plt.subplot(1, 3, 3)
plot_deciles_of_average_tx(fig_nucleosomes_deciles, 'entropy', np.mean)
plt.suptitle("Average expression level")
plt.tight_layout()

plt.figure(figsize=(9, 4))
plt.subplot(1, 3, 1)
plot_deciles_of_average_tx(fig_nucleosomes_deciles, 'positioning', np.std)
plt.ylabel("Cyclicity decile")
plt.xlabel("Average $\\sigma$")

plt.subplot(1, 3, 2)
plot_deciles_of_average_tx(fig_nucleosomes_deciles, 'occupancy', np.std)
plt.xlabel("Average $\\sigma$")

plt.subplot(1, 3, 3)
plot_deciles_of_average_tx(fig_nucleosomes_deciles, 'entropy', np.std)
plt.xlabel("Average $\\sigma$")
plt.suptitle("Average transcriptional variation")
plt.tight_layout()
```
