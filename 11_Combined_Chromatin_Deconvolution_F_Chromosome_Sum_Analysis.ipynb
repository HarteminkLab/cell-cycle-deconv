# Deconvolved Chromosome Window Analysis

In the combined chromatin deconvolution results. We will aim to see how well the deconvolved F images match up to the literature on origin of replication timing from the Brewer lab, 2001.

1. For Chromosome 6.
2. Collect all Gene F images
3. For each F timepoint, slide a 10kb window across the chromosome and sum the reads for each F gene in the window. Take the average if there are overlapping gene windows.
4. Construct a heatmap for all timepoints
5. Compare to the origin timing profiles depicted in Brewer.
